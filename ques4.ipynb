```python
import nltk
from nltk.corpus import movie_reviews
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('movie_reviews')
```

```python
docs=[]
labels=[]
for i in movie_reviews.fileids():
    docs.append(" ".join(movie_reviews.words(i)))
    labels.append(movie_reviews.categories(i)[0])
```

```python
print(docs[0])
print()
print(labels[0])
```

```python
print(docs[1001])
print()
print(labels[1001])
```

```python
X_train,X_test,y_train,y_test = train_test_split(docs, labels, test_size=0.2, random_state=42)
```

```python
tfidf = TfidfVectorizer(stop_words="english", max_features=5000)

X_train = tfidf.fit_transform(X_train)
X_test = tfidf.transform(X_test)
```

```python
clf = LogisticRegression(max_iter=1000)
clf.fit(X_train,y_train)
```

```python
pred = clf.predict(X_test)

print("Accuracy:",accuracy_score(y_test,pred))
```

```python
t=["This movie was really amazing and I loved it!"]

print("Custom:",clf.predict(tfidf.transform(t))[0])
```
