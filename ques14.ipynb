```python
import os
os.environ["USE_TF"] = "0"
```

```python
from sentence_transformers import SentenceTransformer, util
from PIL import Image
```

```python
model = SentenceTransformer("clip-ViT-B-32")

image_paths = ["dog.jpg", "class.jpg"]
images = [Image.open(p).convert("RGB") for p in image_paths]
```

```python
captions = [
    "A dog playing on the grass.",
    "Students sitting in a classroom.",
    "A plate of food on a table."
]
```

```python
image_embs = model.encode(images, convert_to_tensor=True, normalize_embeddings=True)
text_embs = model.encode(captions, convert_to_tensor=True, normalize_embeddings=True)

sim = util.cos_sim(image_embs, text_embs)
```

```python
for i,p in enumerate(image_paths):
    best=int(sim[i].argmax())

    print("\nImage:",p)
    print("Best caption:",captions[best])
    print("Similarity:",float(sim[i][best]))
```

```python
def is_mismatch(i,j,thr=0.25):
    return float(sim[i][j])<thr
```

```python
test_img,test_cap=0,2

print("\nCheck mismatch:",captions[test_cap])
print("Similarity:",float(sim[test_img][test_cap]))
print("Mismatch?",is_mismatch(test_img,test_cap))
```
