```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad", framework="pt")

context = "The Taj Mahal is located in Agra and was built by Shah Jahan."

question = "Where is the Taj Mahal located?"

res = qa(question=question, context=context)

print("answer:", res["answer"], "score:", round(res["score"],3))
```
