```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from collections import Counter
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("punkt_tab")
```

```python
text = open("data.txt","r").read().lower()

tokens=[]
for i in word_tokenize(text):
    if i.isalpha():
        tokens.append(i)
        
sw = set(stopwords.words("english"))
sw_tokens=[]
for i in tokens:
    if i not in sw:
        sw_tokens.append(i)

ps = PorterStemmer()
stems=[]
for i in sw_tokens:
    stems.append(ps.stem(i))
    
lm = WordNetLemmatizer()
lemms=[]
for i in sw_tokens:
    lemms.append(lm.lemmatize(i))
```

```python
print("Top 10 tokens:")
for i in Counter(tokens).most_common(10):
    print(i)
print()
    
print("Top 10 stems :") 
for i in Counter(stems).most_common(10):
    print(i)
print()

print("Top 10 lemmas:") 
for i in Counter(lems).most_common(10):
    print(i)
```

```python
wc = WordCloud().generate(" ".join(tokens))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```
