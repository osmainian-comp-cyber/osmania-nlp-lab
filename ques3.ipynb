```python
import nltk
import math
import random
from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
text = open("data.txt").read().lower()

sents = []
for i in sent_tokenize(text):
    s=[]
    s.append("<s>")
    s.extend(word_tokenize(i))
    s.append("</s>")
    sents.append(s)
```

```python
uni = {}
bi = {}

for s in sents:
    for i in range(len(s)):
        if s[i] in uni:
            uni[s[i]] += 1
        else:
            uni[s[i]] = 1
        if i > 0:
            pair=(s[i-1],s[i])
            if pair in bi:
                bi[pair]+=1
            else:
                bi[pair]=1

V = len(uni)
T = sum(uni.values())

def Pu(w):
    return (uni[w]+1)/(T+V)

def Pb(w1,w2):
    return (bi[(w1,w2)]+1)/(uni[w1]+V)

def ppl(model):
    LP = 0
    N = 0
    for s in sents:
        for i in range(1,len(s)):
            if model == "uni":
                p = Pu(s[i])
            else:
                p = Pb(s[i-1],s[i])
            LP += math.log(p)
            N += 1

    return math.exp(-LP/N)

print("Unigram PPL:",ppl("uni"))
print("Bigram PPL:",ppl("bi"))
```

```python
def gen():
    w = "<s>"
    out = []
    for i in range(20):
        nxt = []
        for pair in bi:
            if pair[0] == w:
                nxt.append(pair[1])
        if len(nxt) == 0:
            break
        w = random.choice(nxt)
        if w == "</s>":
            break
        out.append(w)
    return " ".join(out)

print("\nGenerated:")
for i in range(3):
    print(gen())
```
